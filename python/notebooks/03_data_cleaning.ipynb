```python
# ============================================================
# Import Libraries
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 20)
```

```python
# ============================================================
# Load Extracted CSV Files
# ============================================================

from pathlib import Path

data_path = Path("../data/extracted")

customers_df = pd.read_csv(data_path / "customers.csv")
products_df = pd.read_csv(data_path / "products.csv")
orders_df = pd.read_csv(data_path / "orders.csv")
suppliers_df = pd.read_csv(data_path / "suppliers.csv")
inventory_df = pd.read_csv(data_path / "inventory.csv")
transportation_df = pd.read_csv(data_path / "transportation_logistics.csv")
payments_df = pd.read_csv(data_path / "payments.csv")
returns_df = pd.read_csv(data_path / "returns.csv")
customer_discounts_df = pd.read_csv(data_path / "customer_discounts.csv")

print("✅ All extracted datasets loaded successfully!")
```

```python
# ============================================================
# Dataset Summary
# ============================================================

datasets = {
    "Customers": customers_df,
    "Products": products_df,
    "Orders": orders_df,
    "Suppliers": suppliers_df,
    "Inventory": inventory_df,
    "Transportation": transportation_df,
    "Payments": payments_df,
    "Returns": returns_df,
    "Customer Discounts": customer_discounts_df
}

summary = []

for name, df in datasets.items():
    summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
# ============================================================
# Missing Value Analysis
# ============================================================

missing_summary = []

for name, df in datasets.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if len(missing) > 0:
        for column, value in missing.items():
            missing_summary.append({
                "Dataset": name,
                "Column": column,
                "Missing Values": value,
                "Percentage": round((value / len(df)) * 100, 2)
            })

missing_df = pd.DataFrame(missing_summary)

missing_df
```

## Missing Value Findings

### Orders

- `postal_code` contains **11 missing values (0.11%)**
- These values were intentionally left as NULL because verified postal codes were unavailable.

### Transportation

- `actual_delivery_date` contains **953 missing values (9.72%)**
- This is expected because some shipments have not yet been delivered.
- Missing values are retained as NULL.

```python
# ============================================================
# Convert Date Columns
# ============================================================

# Orders Table
orders_df["order_date"] = pd.to_datetime(orders_df["order_date"])
orders_df["ship_date"] = pd.to_datetime(orders_df["ship_date"])
orders_df["created_at"] = pd.to_datetime(orders_df["created_at"])

# Transportation Table
transportation_df["dispatch_date"] = pd.to_datetime(
    transportation_df["dispatch_date"]
)

transportation_df["estimated_delivery_date"] = pd.to_datetime(
    transportation_df["estimated_delivery_date"]
)

transportation_df["actual_delivery_date"] = pd.to_datetime(
    transportation_df["actual_delivery_date"]
)

transportation_df["created_at"] = pd.to_datetime(
    transportation_df["created_at"]
)

print("✅ Date columns converted successfully!")
```

```python
# ============================================================
# Duplicate Analysis
# ============================================================

duplicate_summary = []

for name, df in datasets.items():
    duplicate_summary.append({
        "Dataset": name,
        "Duplicate Rows": df.duplicated().sum()
    })

duplicate_df = pd.DataFrame(duplicate_summary)

duplicate_df
```

```python
# ============================================================
# Verify Data Types
# ============================================================

print("Orders Table")
print(orders_df.dtypes)

print("\n" + "="*50 + "\n")

print("Transportation Table")
print(transportation_df.dtypes)
```

## Data Type Conversion

### Objective

Convert all date-related columns from text (`object`) to proper `datetime` format.

### Converted Columns

**Orders**
- order_date
- ship_date
- created_at

**Transportation**
- dispatch_date
- estimated_delivery_date
- actual_delivery_date
- created_at

### Result

All date columns were successfully converted to `datetime64[ns]`, making them suitable for time-series analysis, feature engineering, and Power BI visualizations.

```python
# ============================================================
# Text Standardization
# ============================================================

# ---------- Orders ----------
text_columns_orders = [
    "country",
    "city",
    "state",
    "region",
    "ship_mode"
]

for col in text_columns_orders:
    orders_df[col] = (
        orders_df[col]
        .astype(str)
        .str.strip()
        .str.title()
    )

# ---------- Transportation ----------
text_columns_transportation = [
    "carrier_name",
    "shipment_status"
]

for col in text_columns_transportation:
    transportation_df[col] = (
        transportation_df[col]
        .astype(str)
        .str.strip()
        .str.title()
    )

print("✅ Text standardization completed successfully!")
```

```python
# ============================================================
# Data Validation Report
# ============================================================

validation_report = []

for name, df in datasets.items():

    validation_report.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

validation_df = pd.DataFrame(validation_report)

validation_df
```

```python
# ============================================================
# Save Cleaned Datasets
# ============================================================

from pathlib import Path

processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)

customers_df.to_csv(processed_path / "customers.csv", index=False)
products_df.to_csv(processed_path / "products.csv", index=False)
orders_df.to_csv(processed_path / "orders.csv", index=False)
suppliers_df.to_csv(processed_path / "suppliers.csv", index=False)
inventory_df.to_csv(processed_path / "inventory.csv", index=False)
transportation_df.to_csv(processed_path / "transportation.csv", index=False)
payments_df.to_csv(processed_path / "payments.csv", index=False)
returns_df.to_csv(processed_path / "returns.csv", index=False)
customer_discounts_df.to_csv(processed_path / "customer_discounts.csv", index=False)

print("===================================================")
print("✅ All cleaned datasets saved successfully!")
print("===================================================")
print(processed_path.resolve())
```

```python
# ============================================================
# Export Cleaned Data
# Enterprise Retail Intelligence Platform
# ============================================================

from pathlib import Path

# Create output directory if it does not exist
processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)

# Dictionary containing cleaned DataFrames
processed_datasets = {
    "customers": customers_df,
    "products": products_df,
    "orders": orders_df,
    "suppliers": suppliers_df,
    "inventory": inventory_df,
    "transportation": transportation_df,
    "payments": payments_df,
    "returns": returns_df,
    "customer_discounts": customer_discounts_df,
}

# Export each dataset
for name, dataframe in processed_datasets.items():
    dataframe.to_csv(processed_path / f"{name}.csv", index=False)

print("=" * 60)
print("SUCCESS: All cleaned datasets exported successfully.")
print("=" * 60)
print(f"Output Location : {processed_path.resolve()}")
print(f"Datasets Exported : {len(processed_datasets)}")
```
