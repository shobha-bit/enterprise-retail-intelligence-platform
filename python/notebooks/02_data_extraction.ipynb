```python
# ============================================================
# Import Required Libraries
# ============================================================

import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
import os
```

```python
# ============================================================
# Load Environment Variables
# ============================================================

load_dotenv()

DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")

print("Environment variables loaded successfully!")
```

```python
# ============================================================
# Create Database Engine
# ============================================================

DATABASE_URL = (
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}"
    f"@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

engine = create_engine(DATABASE_URL)

print("Database Engine Created Successfully!")
```

```python
# ============================================================
# Test Database Connection
# ============================================================

try:
    with engine.connect() as connection:
        print("✅ PostgreSQL Connection Successful!")
except Exception as e:
    print("❌ Connection Failed")
    print(e)
```

```python
# ============================================================
# Reusable Function to Load Any Database Table
# ============================================================

def load_table(table_name):
    """
    Load a PostgreSQL table into a Pandas DataFrame.
    """

    query = f"SELECT * FROM {table_name};"

    df = pd.read_sql(query, engine)

    print(f"✅ {table_name} loaded successfully - {df.shape[0]} rows, {df.shape[1]} columns")

    return df
```

```python
# ============================================================
# Extract All Database Tables
# ============================================================

customers_df = load_table("customers")
products_df = load_table("products")
orders_df = load_table("orders")
suppliers_df = load_table("suppliers")
inventory_df = load_table("inventory")
transportation_df = load_table("transportation_logistics")
payments_df = load_table("payments")
returns_df = load_table("returns")
customer_discounts_df = load_table("customer_discounts")
```

```python
# ============================================================
# Validation Summary
# ============================================================

table_summary = pd.DataFrame({
    "Table": [
        "Customers",
        "Products",
        "Orders",
        "Suppliers",
        "Inventory",
        "Transportation",
        "Payments",
        "Returns",
        "Customer Discounts"
    ],
    "Rows": [
        customers_df.shape[0],
        products_df.shape[0],
        orders_df.shape[0],
        suppliers_df.shape[0],
        inventory_df.shape[0],
        transportation_df.shape[0],
        payments_df.shape[0],
        returns_df.shape[0],
        customer_discounts_df.shape[0]
    ],
    "Columns": [
        customers_df.shape[1],
        products_df.shape[1],
        orders_df.shape[1],
        suppliers_df.shape[1],
        inventory_df.shape[1],
        transportation_df.shape[1],
        payments_df.shape[1],
        returns_df.shape[1],
        customer_discounts_df.shape[1]
    ]
})

print("=" * 60)
print("DATA EXTRACTION SUMMARY")
print("=" * 60)

display(table_summary)
```

```python
from pathlib import Path

output_path = Path("../data/extracted")

output_path.mkdir(parents=True, exist_ok=True)

print(output_path.resolve())
```

```python
# ============================================================
# Export All Tables to CSV
# ============================================================

from pathlib import Path

output_path = Path("../data/extracted")
output_path.mkdir(parents=True, exist_ok=True)

customers_df.to_csv(output_path / "customers.csv", index=False)
products_df.to_csv(output_path / "products.csv", index=False)
orders_df.to_csv(output_path / "orders.csv", index=False)
suppliers_df.to_csv(output_path / "suppliers.csv", index=False)
inventory_df.to_csv(output_path / "inventory.csv", index=False)
transportation_df.to_csv(output_path / "transportation_logistics.csv", index=False)
payments_df.to_csv(output_path / "payments.csv", index=False)
returns_df.to_csv(output_path / "returns.csv", index=False)
customer_discounts_df.to_csv(output_path / "customer_discounts.csv", index=False)

print("✅ All CSV files exported successfully!")
print(output_path.resolve())
```
